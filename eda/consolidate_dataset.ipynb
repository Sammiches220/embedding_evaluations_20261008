```python
import pandas as pd
from pathlib import Path
import numpy as np

SAMPLE_DATA_PARQUET_PATH = Path.cwd().parent / "data" / "posts-sample.parquet"

CLUSTER_DIR = Path.cwd().parent / "data" / "clusters"
CANONICAL_CLUSTER_DIR = CLUSTER_DIR / "canonical"
ARTIST_AVG_FAV_COUNT_CLUSTER = CANONICAL_CLUSTER_DIR / "artist_avg_fav_count" / "labels.npy"

DANBOORU_ITEM_MAPPING = Path.cwd().parent / "data" / "danbooru_eda" / "danbooru_item_mapping.npy"
DOMAIN_POST_IDS_PATH = Path.cwd().parent / "data" / "danbooru_eda" / "domain_post_ids.npy"

EMBEDDINGS_PATH = Path("/home/main/Desktop/files/projects/hycompass/data/danbooru_input_v1/embeddings.npy")
```

```python
## Load data
sample_df = pd.read_parquet(SAMPLE_DATA_PARQUET_PATH)  ## Needs pandas, pyarrow upgraded python -m pip install --upgrade pandas pyarrow
artist_avg_fav_count = np.load(ARTIST_AVG_FAV_COUNT_CLUSTER)
# danbooru_item_mapping = np.load(DANBOORU_ITEM_MAPPING)
danbooru_domain_post_ids = np.load(DOMAIN_POST_IDS_PATH)
```

```python
# Build the lookup once
domain_index = pd.Index(danbooru_domain_post_ids.astype(str))
positions = domain_index.get_indexer(sample_df["id"].astype(str))
valid = positions >= 0

# Add all cluster files as columns
for path in CANONICAL_CLUSTER_DIR.glob("*/labels.npy"):
    labels = np.load(path)

    result = np.zeros(len(sample_df), dtype=labels.dtype)
    result[valid] = labels[positions[valid]]

    sample_df[f"{path.parent.name}_cluster"] = result
```

```python
# Memory-map the 30 GB file (doesn't load everything)
embeddings = np.load(EMBEDDINGS_PATH, mmap_mode="r")

# Retrieve only embeddings corresponding to sample posts
sample_embeddings = np.full(
    (len(sample_df), embeddings.shape[1]),
    np.nan,
    dtype=np.float32
)

sample_embeddings[valid] = embeddings[positions[valid]]

# Add embeddings to DataFrame
sample_df["embedding"] = list(sample_embeddings)
```
